11. Container With Most Water


```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        left = 0
        right = len(height) - 1
        max_area = 0

        while left < right:
            width = right - left
            ht = min(height[left], height[right])
            area = width * ht
            max_area = max(max_area, area)

            # Move the shorter wall inward
            if height[left] < height[right]:
                left += 1
            else:
                right -= 1

        return max_area
```
