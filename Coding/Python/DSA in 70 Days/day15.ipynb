# text editor using linked list 

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
```

```python
class TextEditor:
    def __init__(self):
        self.head = None
        self.history = []  # For undo operations

    # Add new word at the end
    def add_text(self, word):
        new_node = Node(word)
        self.history.append(("ADD", word))  # Save history for undo

        if not self.head:  # First word
            self.head = new_node
        else:
            temp = self.head
            while temp.next:
                temp = temp.next
            temp.next = new_node

    # Display complete text
    def display_text(self):
        if not self.head:
            print("Document is empty!")
            return

        temp = self.head
        text = []
        while temp:
            text.append(temp.data)
            temp = temp.next
        print("📄 Document:", " ".join(text))

    # Undo last action (remove last word)
    def undo(self):
        if not self.history:
            print("Nothing to undo!")
            return

        action, word = self.history.pop()
        if action == "ADD":
            # Remove last node
            if self.head and not self.head.next:
                self.head = None
            else:
                temp = self.head
                while temp.next and temp.next.next:
                    temp = temp.next
                temp.next = None
            print(f"⏪ Undo: Removed '{word}'")

    # Find the middle word using fast & slow pointers
    def find_middle_word(self):
        if not self.head:
            print("Document is empty!")
            return

        slow = self.head
        fast = self.head
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

        print(f"🔹 Middle word: '{slow.data}'")
```

```python
editor = TextEditor()


editor.add_text("Hello")
editor.add_text("world")
editor.add_text("this")
editor.add_text("is")
editor.add_text("Python")
editor.add_text("editor")

# Display full text
editor.display_text()  


# Find middle word
editor.find_middle_word()  



editor.undo()
editor.display_text()  

```
