What is an Array?

An array is a collection of elements stored in contiguous memory locations, usually of the same data type.

```python
# Declaration in Python
arr = [10, 20, 70, 40]  # an array (technically, a list in Python)

# Accessing elements
print(arr[2])  

# Length of array
print(len(arr))  # 4
```

Find the number element in an array

Option 1: Loop over values directly
python

```python
arr = [2,5,6,78,8,86,53]

def findmax(arr):
    max_num = arr[0]
    for num in arr:
        if num>max_num:
            max_num = num
    return max_num
print(findmax(arr))

```

Option 2: If you want to use index:

```python
arr = [2,5,6,78,8,86,53]

def findmax(arr):
    max_num = arr[0]
    for i in range(1,len(arr)):
        if arr[i] > max_num:
            max_num = arr[i]
    return max_num
print(findmax(arr))
```

When to Loop Over Values vs Indexes in an Array

Loop over values when:
You just need to access or compare values, and don’t care about the position.

Use When:
<br>
Finding max/min
<br>
Printing all elements
<br>
Summing numbers
<br>
Searching for a value

```python
for num in arr:
    print(num)
```

Loop over indexes when:

You need the position (index) for:
<br>
Changing values
<br>
Comparing with neighbors (like arr[i+1])
<br>
Tracking index as part of result (e.g., in Two Sum)

 Use When:<br>
Modifying values (arr[i] = ...)
<br>
Accessing arr[i+1], arr[i-1]
<br>
Returning the index as output



```python
for i in range(len(arr)):
    print(f"Index {i} = Value {arr[i]}")
```

| Goal                   | Use                              |
| ---------------------- | -------------------------------- |
| I need values only     | `for num in arr`                 |
| I need index and value | `for i in range(len(arr))`       |
| I want both easily     | `for i, num in enumerate(arr)`  |


Replace even values with 0

```python
arr = [3, 4, 2, 5, 88, 99, 0, 8, 7, 7]

for i in range(len(arr)):
    if arr[i] % 2 == 0:   # Check if even
        arr[i] = 0        # Replace with 0

print(arr)
```

Alternative (with enumerate)

```python

for i, val in enumerate(arr):
    if val % 2 == 0:
        arr[i] = 0
print(arr)
```

 Problem 2: Reverse an array

```python
arr = [2,3,4,5,6,7,8,9,10]

def reverse_arr(arr):
    return arr[::-1]

print(reverse_arr(arr))  # ✅ Correct! Function call + print result
# Output: [10, 9, 8, 7, 6, 5, 4, 3, 2]
```

Approach 1: Brute Force (O(n²))

```python
nums = [2,7,11,15]
target=12
def twosum(nums,target):
    for i in range(0,len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i]+nums[j]==target:
                return [i,j]
print(twosum(nums,target))
        
```

Hash Map (O(n) time):

```python

nums = [2,7,11,15]
target=9
def twosum(nums,target):
    hash_map={}
    for i ,num in enumerate(nums):
        complement = target - num
        if complement in hash_map:
            return [hash_map[complement],i]
    
        hash_map[num] = i
    return None 
print(twosum(nums,target))
```

Two-pointer approach (O(n log n) time):

```python
nums = [2,7,11,15]
target=9
def twoSum(nums, target):

    # Create list of (value, original index) pairs
    nums_with_index = [(num, i) for i, num in enumerate(nums)]
    nums_with_index.sort()  # Sort based on values
    
    left, right = 0, len(nums)-1
    while left < right:
        current = nums_with_index[left][0] + nums_with_index[right][0]
        if current == target:
            return [nums_with_index[left][1], nums_with_index[right][1]]  # Return original indices
        elif current < target:
            left += 1
        else:
            right -= 1

print(twosum(nums,target))
```
