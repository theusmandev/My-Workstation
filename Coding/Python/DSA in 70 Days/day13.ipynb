```python
def containsDuplicate(nums):
    seen = set()
    for n in nums:
        if n in seen:
            return True
        else:
            seen.add(n)
    return False

print(containsDuplicate([1,2,4,3,1]))
print(containsDuplicate([1,2,3,4]))
```

```python
def longestConsecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        # start only if it's the first number in a sequence
        if num - 1 not in num_set:
            length = 1
            while num + length in num_set:
                length += 1
            longest = max(longest, length)

    return longest

# Example
print(longestConsecutive([100,4,200,1,3,2]))  # 4 -> (1,2,3,4)

```
