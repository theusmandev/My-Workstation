```python
def maxArea(height):
    left,right=0,len(height)-1
    max_water = 0
    while left<right:
        width = right - left
        area = min(height[left],height[right])*width
        max_water = max(max_water,area)

        if height[left]<height[right]:
            left+=1
        else:
            right-=1
    return max_water
print("Maximum area:",maxArea(height))

```

```python
def maxArea(height):
    left, right = 0, len(height) - 1
    max_water = 0
    max_left = max_right = 0  # To store the best 

    while left < right:
        width = right - left
        area = min(height[left], height[right]) * width

        # If new max is found, store the indices
        if area > max_water:
            max_water = area
            max_left, max_right = left, right

        # Move the pointer from the smaller height
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    # Final output
    print(f"Maximum area: {max_water}")
    print(f"Between indices: {max_left} and {max_right}")
    print(f"Distance between lines: {max_right - max_left}")
    print(f"Heights: {height[max_left]} and {height[max_right]}")
    return max_water

# Example input
height = [1,8,6,2,5,4,8,3,7]
maxArea(height)
```

```python
import matplotlib.pyplot as plt

def maxArea(height):
    left, right = 0, len(height) - 1
    max_water = 0
    max_left = max_right = 0

    while left < right:
        width = right - left
        area = min(height[left], height[right]) * width

        if area > max_water:
            max_water = area
            max_left, max_right = left, right

        if height[left] < height[right]:
            left += 1
        else:
            right -= 1

    # Plotting
    x = list(range(len(height)))
    y = height

    plt.figure(figsize=(10, 6))
    
    # Draw all lines in blue
    for i in range(len(height)):
        plt.plot([x[i], x[i]], [0, y[i]], color='blue')

    # Highlight the max area container in red
    plt.plot([max_left, max_left], [0, height[max_left]], color='red', linewidth=3)
    plt.plot([max_right, max_right], [0, height[max_right]], color='red', linewidth=3)

    # Shade the area (optional)
    width = max_right - max_left
    min_height = min(height[max_left], height[max_right])
    plt.fill_between([max_left, max_right], min_height, color='skyblue', alpha=0.3)

    plt.title(f"Max Area = {max_water} (Between index {max_left} and {max_right})")
    plt.xlabel("Index")
    plt.ylabel("Height")
    plt.grid(True)
    plt.show()

    return max_water

# Example input
height = [1,8,6,2,5,4,8,3,7]
maxArea(height)
```
