```python
# Step 1: Custom for loop definition
def custom_for(iterable, function):
    iterator = iter(iterable)
    while True:
        try:
            item = next(iterator)
            function(item)
        except StopIteration:
            break

# Step 2: Custom function to apply on each item
def process_log_line(line):
    print(f"[LOG] {line.upper()}")

# Step 3: Sample iterable data (simulated log lines)
log_lines = [
    "user logged in",
    "error connecting to server",
    "file uploaded successfully",
    "user logged out"
]

# Step 4: Use the custom_for loop
custom_for(log_lines, process_log_line)
```

```python
# Step 1: Custom for loop using a for-loop internally
def custom_for(iterable, function):
    for item in iterable:
        function(item)

# Step 2: Custom function to apply on each item
def process_log_line(line):
    print(f"[LOG] {line.upper()}")

# Step 3: Sample iterable data (simulated log lines)
log_lines = [
    "user logged in",
    "error connecting to server",
    "file uploaded successfully",
    "user logged out"
]

# Step 4: Use the custom_for loop
custom_for(log_lines, process_log_line)
```
